```python
### extracting ratings of people who rated atleast 200 books
ratings = ratings[ratings["user_id"].isin(y)]
```

```python
### merge ratings and books
ratings_with_books = ratings.merge(books, on = "ISBN")
```

```python
### Group the rows by book title, then count how many ratings each book has received.
ratings_with_books.groupby('title')['rating'].count()
```

```python
number_rating = ratings_with_books.groupby('title')['rating'].count().reset_index()
```

```python
number_rating.head(2)
```

```python
number_rating.rename(columns={
    "rating" : "number of ratings"
}, inplace=True)
```

```python
number_rating.head(2)
```

```python
number_rating.shape
```

```python
### merging ratings_with_books and number_rating
final_rating = ratings_with_books.merge(number_rating, on='title')
```

```python
final_rating.head(2)
```

```python
final_rating['number of ratings'] >= 50
```

```python
final_rating = final_rating[final_rating['number of ratings'] >= 50]
```

```python
final_rating.shape
```

```python
final_rating.drop_duplicates(['user_id', 'title'], inplace=True)
```

```python
final_rating.shape
```

```python
book_pivot = final_rating.pivot_table(columns='user_id', index='title', values='rating')
```

```python
book_pivot.shape
```

```python
book_pivot
```

```python
book_pivot.fillna(0, inplace=True)
```

```python
# Ratings
#    ↓
# final_rating
#    ↓
# pivot_table()
#    ↓
# Book × User matrix
#    ↓
# Similarity calculation
#    ↓
# Find similar books
#    ↓
# Recommendations
```

```python
from scipy.sparse import csr_matrix
book_sparse = csr_matrix(book_pivot)
```

```python
# csr_matrix : it's useful because your book_pivot matrix will contain a huge number of zeros/missing values.
# CSR means :
# Compressed Sparse Row

# Instead of storing every 0, it mainly stores the non-zero values and their locations.
```

```python
print(book_sparse)
# book_sparse.shape
```

```python
type(book_sparse)
```

```python
from sklearn.neighbors import NearestNeighbors
model = NearestNeighbors(algorithm='brute')
```

```python
model.fit(book_sparse)
```

```python
book_pivot.iloc[237].values.reshape(1,-1).shape
```

```python
distances, suggestions = model.kneighbors(book_pivot.iloc[237].values.reshape(1,-1), n_neighbors=6)
```

### Function to recommend books.

```python
def recommend_book(book_name):
    book_id = np.where(book_pivot.index == book_name)[0][0]
    distances, suggestions = model.kneighbors(book_pivot.iloc[book_id].values.reshape(1,-1), n_neighbors=6)
    print(f"The suggestions for {book_name} are :")
    for i in range(len(suggestions[0])):  # here suggestions is a 2D array
        if i!=0:
            print(f"{i}. {book_pivot.index[suggestions[0][i]]}")
```

```python
recommend_book('Animal Farm')
```

```python
recommend_book('Harry Potter and the Chamber of Secrets (Book 2)')
```
